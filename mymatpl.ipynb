```python
rice_crop=[12,15,5,21,24]
crop_year=[2012,2015,2018,2021,2024]
wheat_crop=[5,10,20,4,4]
plt.plot(crop_year,rice_crop,"s--b");
plt.plot(crop_year,wheat_crop,'*-r');
plt.legend(["rice","wheat"])
```

```python
#to change the figure size to our own custom
plt.figure(figsize=(5,5))
```

```python
sns.set_style("whitegrid")
plt.plot(crop_year,rice_crop,"s-b")
```

```python
sns.set_style("darkgrid")
plt.plot(crop_year,wheat_crop,"*--r");
```
