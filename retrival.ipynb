```python
import pandas as pd
import numpy as np
```

```python
my_data={
    "NAME":["BALAJI","BHAGIRATH","GAGAN","ABHI","RAM"],
    "MARKS":[75,85,95,45,56],
    "HOURS":[4,3,2,1,5]

}
df=pd.DataFrame(my_data)
print(df)
```

```python
lessscore=df[df["MARKS"]>45]
print(lessscore)
```

```python
studyhours=df[df["HOURS"]>2].sort_values("HOURS",ascending=True)
print(studyhours)
```

```python
result=df["PASS"]=df["MARKS"]>=55
print(df)


```

```python
df["PASS"]=np.where(df["MARKS"]>=55,"PASSED","FAILED")
print(df)
```

```python
df["LEVELS"] = np.select(
    [
        df["HOURS"] > 4,
        df["HOURS"] >= 3
    ],
    [
        "high",
        "medium"
    ],
    default="low"
)
print(df)
```

Here are the **short definitions only** for what you learned today:

* **DataFrame:** A two-dimensional table in Pandas containing rows and columns.

* **CSV:** A text file format used to store tabular data, where values are separated by commas.

* **Querying/Filtering:** Selecting only the rows that satisfy a given condition.

* **Boolean condition:** A condition that produces `True` or `False` for each row.

* **Sorting:** Arranging DataFrame rows in a particular order based on one or more columns.

* **`query()`:** A Pandas method used to filter rows using a condition written as a query expression.

* **Adding a column:** Creating a new column in a DataFrame using existing or new data.

* **`np.where()`:** Selects one of two values depending on whether a condition is `True` or `False`.

* **`np.select()`:** Selects values based on multiple conditions, using the first condition that is `True`.

* **`drop()`:** Removes specified rows or columns from a DataFrame.

* **`NaN`:** Represents a missing or unavailable value in a dataset.

* **`isna()`:** Identifies missing values in a DataFrame.

* **`fillna()`:** Replaces missing values with a specified value or calculated value.

* **`axis=0`:** Refers to rows.

* **`axis=1`:** Refers to columns.
