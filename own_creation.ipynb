HOW TO CREATE THE A FILE AND CREATE THE FAKE DATE FOR US TO PRATICE THE PYTHON PANDAS 

```python
import pandas as pd
import numpy as np
np.random.seed(42)
dates=pd.date_range(start="2020-03-01",periods=900,freq="D")
new_cases=np.random.randint(1000,500000,size=900)
new_deaths=np.random.randint(1000,20000,size=900)
new_tets=np.random.randint(100000,200000,size=900)
covid_data_own=pd.DataFrame({"date":dates,"new_cases":new_cases,"new_deaths":new_deaths,"new_tests":new_tets})
covid_data_own.to_csv("balaji_data.csv",index=False)
print(covid_data_own)
print()
print("csv file create:balaji_data.csv")
```
