```python
arr = np.array([5, 10, 15, 20, 25])
print(np.sum(arr))
arr2=np.array([[5,4,6],[8,5,6]])
print(np.sum(arr2))
#axis zero gives the sum of rows
#axis 1 gives the sum of columns
print(np.sum(arr2,axis=0))
print(np.sum(arr2,axis=1))
```

```python
meanarray=np.array([10,20,30,40,50])
#mean prints the average of the given array
print(np.mean(meanarray))
#example for matrix 
arrmean=np.array([[5,4,6],[8,5,6]])
print(np.mean(arrmean))
#axis zero for average of rows
print(np.mean(arrmean,axis=0))
#axis one for average of columns
print(np.mean(arrmean,axis=1))

```
